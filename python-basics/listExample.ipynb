## Example 2: Organizing Student Grades
#### Create a list store and calculate average grades for student 

```python
grades=[76,50,82,95,98]

grades.append(68)

#calculate average grade:
averageGrade=sum(grades)/len(grades)
print(f"The Average Grade is: {averageGrade}")

#highest grade:
highestGrade=max(grades)
lowestGrade=min(grades)

print(f"The highest grade is: {highestGrade} and the lowsest grade is: {lowestGrade}")
```

## Example 3: Managing An inventory
#### Use a list to manage inventory items in a store

```python
# Managing An inventory
inventory=["apple", "grapes", "orange", "kiwi"]
inventory.append("mango")

inventory.remove("grapes")

# print(inventory)

# Check if a item is in stock 
item=input("enter to search:")
if item in inventory:
    print(f"{item} are in stock")
else:
    print(f"{item} are in not stock")

# inventory list:
list=[itm for itm in inventory]
print(list)
```
