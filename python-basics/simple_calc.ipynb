```python
### SIMPLE CALCULATOR:

num1 = int(input("enter first name:"))
num2 = int(input("enter second name:"))
operation = input("enter operation (+, -, *, /):")

if operation == "+":
    result = num1 + num2
    print("sum of two number is: ", result)
elif operation == "-":
    result = num1 - num2
    print("difference of two number is: ", result)
elif operation == "*":
    result = num1 * num2
    print("product of two number is : ", result)
elif operation == "/":
    result = num1 / num2
    print("quotient of two number is :", result)


# num1 = int(input("enter first name:"))
# num2 = int(input("enter second name:"))

# sum = num1 + num2
# minus = num1 - num2
# multy = num1 * num2
# divide = num1 / num2

# print("sum of two numbers is:", sum)
# print("difference of two numbers is:", minus)
# print("product of two numbers is:", multy)
# print("quotient of two numbers is:", divide)
```
