## Encapsulation And Abstraction:
#### Encapsulation and abstraction are two fundamental principles of Objects-Orirented Programming (OOPS) that help in designing robust, maintainable and reusable code. Encapsulation involves bunding data and methods that operate on the data within a single unit,while abstraction involves hiding complex implementation details and exposing only necessary features.

## Encapsulation:
#### Encapsulation is the concept of wrapping data (variable) and methods (function) together as a single unit. it restricts direct access to some of the object's components, which is a means of preventing accidental interference and misuse of the data.

```python
### Encapsulation with Getter and Setter Method
### Public, private & protected or access modifiers
## Public code:
class Person: # class
    def __init__(self, name, age):  # constructor

        self.name=name # public variables
        self.age=age   # public variables

def get_name(person):
    return person.name

# creating objects:
pr=Person("shuvadeep", 30)
# print(pr.name)
dir(pr)
get_name(pr)


```

```python
## Protected code:

class Person: # class
    def __init__(self, name, age, gender):  # constructor

        self._name=name         # protected variables
        self._age=age           # protected variables
        self.gender=gender      # public variables

# Derived class:
class Employee(Person):  # inherting the Person()
    def __init__(self, name, age, gender):
        super().__init__(name, age, gender)

emp=Employee("shuvadeep", 30, "male")
print(emp._name)
```

```python
## Encapsulation With Getter and Setter

class Person:
    def __init__(self, name, age):
        self.__name=name # Private access modifier or variable
        self.__age=age   # Private variable

    ## getter method for name:
    def get_name(self):
        return self.__name

    ## setter method for name:
    def set_name(self, name):
        self.__name=name

person=Person("shuvadeep", 30)

# Access and modify private variables using getter and setter

print(person.get_name())
person.set_name("shuvadeep podder")
print(person.get_name())
        
```
