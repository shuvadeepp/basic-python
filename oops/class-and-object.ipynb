```python
## Instance variable and methods:
# created class:
class Dog:
    ## contructor 
    def __init__(self, name, age):
        self.name=name
        self.age=age 

# created object:
dogOne=Dog("tom", 2)
print(f"dog name is {dogOne.name} and the age is {dogOne.age}") 
```

```python
class BankAcount:
    def __init__(self, owner, balance=0):
        self.owner=owner
        self.balance=balance

    def deposit(self, amount):
        self.balance+=amount
        print(f"{amount} is deposit. New balance is {self.balance}")

    def withdraw(self, amount):
        if amount>self.balance:
            print("Insufficient balance")
        else:
            self.balance-=amount
            print(f"{amount} is withdrawn. New balance is {self.balance}")

    def get_balance(self):
        # print("amt", self.balance)
        # return self.balance
        if self.balance<=0:
            print("you dont have any balance.")
        else:
            print(f"Your account balance is {self.balance}")

''' create an account: ''' 

userAccOne=BankAcount("shuvadeep", 500)
print(userAccOne.balance)
```
