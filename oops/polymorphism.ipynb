## Polymorphism
#### Polymorphism is a core concept in Object-Oriented programming (OOPS) that allows object of different classes to be treated as objects of a common superclass. it provides a way to perform a single action in different forms. Polymorphism is typically achieved through overriding and interfaces

## Method Overrinding:
#### Method overriding allows a child to provide a specific implementation of a method that is already defined in its parent class

```python
# Base class
class Animal:
    def speak(self):
        return "Sound of the animal"

# derived class 1
class Dog(Animal):
    def speak(self):
        return "woof!"

# derived class 2
class Cat(Animal):
    def speak(self):
        return "meow!"

# dog=Dog()
# cat=Cat()
# print(dog.speak())
# print(cat.speak())

animals=[Dog(), Cat()]
for animal in animals:
    print(animal.speak()) 
```

```python
### Polymorphism with function and methods:
# base class

class Shape:
    def area(self):
        return "the area of the figure"

# Derived class
class Rectangular(Shape):
    def __init__(self, width, height):
        self.width=width
        self.height=height

    def area(self):
        return self.width * self.height

#derived class two
class Circle(Shape):
    def __init__(self, radius):
          self.radius=radius

    def area(self):
        return 3.14 * self.radius * self.radius

# Function that demonstrates polymonphism

def print_area(Shape):
    # print(f"the area is {Shape.area()}")
    return f"the area is {Shape.area()}"

# objects:
rectangular=Rectangular(4,5)
circle=Circle(3)

#printing:
# print(print_area(rectangular))
# print(print_area(circle))

shape = [rectangular, circle]
for shap in shape:
    print(print_area(shap))
```

## Polymorphism with Abstact base classes
#### Abstract base classes (ABCs) are used to define common methods for a group of related objects. The can exforce that derived classes implement particular methods, promoting consistency across different implementations.  

```python
from abc import ABC, abstractmethod

# Define an abstract class

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):    # rule: har child ko ye likhna hi hoga
        pass

# Derived class 1
class Car(Vehicle):
    def start_engine(self):
        return "car engine started"

# Derived class 2:
class Motocycle(Car):
    def start_engine(self):
        return "Motocycle engine started"

# function that demonstrates polymorphism:    
def start_vehicle(vehicleParams):
    print(vehicleParams.start_engine())

## Create objects of car and Motocycle:
car = Car()
motocycle = Motocycle()

v = [car, motocycle]
for i in v:
    start_vehicle(i)

# start_vehicle(car)
# start_vehicle(motocycle)
```
