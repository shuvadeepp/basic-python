## Inheritance in pyhton:
#### Inheritance is a fundamental concept in Object-Otented programming (OOPS) that allows a class to inherit attributes and methods from another class. this lesson covers  single inheritance and multiple ingeritance, demonstrating how to create and use them in python.

```python
# Inheritance (Single inheritance)
# parent class
class Car:
    def __init__(self, windows, doors, engintype):
        self.windows=windows
        self.doors=doors
        self.engintype=engintype

    def drive(self):
        # print(f"the person will drive the {self.engintype} car")
        return f"the person will drive the {self.engintype} car"
```

```python
car1=Car(4,5,"petrol car")
print(car1.drive())
```

```python
class Tesla(Car):
    def __init__(self, windows, doors, engintype, self_driving):

        super().__init__(windows, doors, engintype)

        self.self_driving=self_driving
    
    def selfDriving(self):
        return f"Tesla supports self driving: {self.self_driving}"
```

```python
tesla1=Tesla(4, 5, "electic", True)
print(tesla1.selfDriving())
```

```python
tesla1.drive()
```

```python
## Multiple inheritance
## When a class inheritance from more then one base class.
## base class 1

class Animal:
    def __init__(self, name):
        self.name=name

    def speak(self):
        print("Subclass must implement this method")
```

```python
# base class 2
class Pet:
    def __init__(self, owner):
        self.owner=owner

## Derived class
class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def speak(self):
        return f"{self.name} say woof"

```

```python
#create object:
dog = Dog("buddy", "shuva")
print(dog.speak())   
print(dog.owner)   
```
