## The filter() function in python 
#### The filter out item from a list (or any other iterable) based on condtion

```python
def even(num):
    if num % 2 == 0:
        return True
```

```python
lst=[1,2,3,4,5,6,7,8,9]

res=list(filter(even, lst))
greater_then_five=list(filter(lambda x:x>5, lst))
multipleCondition=list(filter(lambda x:x>5 and x%2==0, lst))

print(res)
print(greater_then_five)
print(multipleCondition)
```

```python
# Apply Filter() to check if the age is greater then 25 in dictionary
data=[
    {"name": "shuvadeep", "age": 30},
    {"name": "bikram", "age": 24},
    {"name": "shima", "age": 35},
]
 
# print(data[0]['age']) 
 
def checkAge(param):
    return param['age']>25

output=list(filter(checkAge, data))
output1=list(filter(lambda x:x['age']>25, data))
print(output)
print(output1)
```
