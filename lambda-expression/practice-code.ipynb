```python
# 1. Sort a list of tuples by second element
students = [("Shuvadeep", 85), ("Priya", 92), ("Amit", 78)]
sort_list=sorted(students, key=lambda x: x[1])
print(sort_list)
```

```python
# 2. Sort list of dictionaries by a key
employees = [
    {"name": "Shuvadeep", "salary": 50000},
    {"name": "Priya", "salary": 70000},
    {"name": "Amit", "salary": 60000}
]

sort_salary=sorted(employees, key=lambda x:x['salary'], reverse=True)
print(sort_salary)
```

```python
# Hardest — group-like operation: filter dict list by multiple conditions, then transform

products = [
    {"name": "Laptop", "price": 50000, "stock": 5},
    {"name": "Mouse", "price": 500, "stock": 0},
    {"name": "Keyboard", "price": 1500, "stock": 10},
    {"name": "Monitor", "price": 12000, "stock": 0}
]
# print(type(products))
availble_stock=list(
    map(
        lambda p: f"{p['name']} - {p['price']}",
        filter(lambda p: p['stock'] > 0 and p['price'] < 20000, products)
    )
)
# print(availble_stock)
available_stock =[]
for prod in products:
    # print(prod['name'], '-' ,prod['price'])
    if prod['stock'] > 0 and prod['price'] < 20000:
        format_item = f"name: {prod['name']} - price: {prod['price']} - qtd: {prod['stock']}"
        available_stock.append(format_item)

print(available_stock)
```
