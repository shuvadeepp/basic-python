```python
# 1. Sort a list of tuples by second element
students = [("Shuvadeep", 85), ("Priya", 92), ("Amit", 78)]
sort_list=sorted(students, key=lambda x: x[1])
print(sort_list)
```

```python
# 2. Sort list of dictionaries by a key
employees = [
    {"name": "Shuvadeep", "salary": 50000},
    {"name": "Priya", "salary": 70000},
    {"name": "Amit", "salary": 60000}
]

sort_salary=sorted(employees, key=lambda x:x['salary'], reverse=True)
print(sort_salary)
```
