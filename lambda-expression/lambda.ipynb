## Lambda function in pyhton
#### Lambda function are small anonymous function defined using the lamda keyword. they can have any number of arguments but only one expression. they are commonly used for short operation or as arguments to higher-order function.

```python
# syntax:
# variable = lambda arguments_name: expression 
```

```python
# normal defination:
def addition(a, b):
    return a + b

# lambda defination:
addition=lambda a, b: a+b
# type(addition) # function
print(addition(5,2))
```

```python
print(addition(5,2))
```

```python
# odd/ even in normal func:
def oddEven(num):
    if num % 2 == 0:
        return "true"
    else:
        return "false"

# print(oddEven(11))

# lambda expression
oddEven=lambda num: "true" if num % 2 == 0 else "false"
print(oddEven(2))
```

```python

def square(numb):
    result=[]
    for i in numb:
        result.append(i**2)
    return result

# print(square([1,2,3,4,5]))

square=lambda numb: list(map(lambda x:x**2, numb))
print(square([1,2,3,4,5]))
```
