```python

# 1. List of dictionaries — filter students who passed (score >= 40), fir unke naam capitalize karo
students = [
    {"name": "shuvadeep", "score": 85},
    {"name": "priya", "score": 32},
    {"name": "amit", "score": 55},
    {"name": "rahul", "score": 20}
]

student_name=list(map(lambda studentName: studentName['name'].upper(), students))
student_pass_score=list(filter(lambda student_score: student_score['score'] > 30, students))

print(student_name)
print(student_pass_score)
```
