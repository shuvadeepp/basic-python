```python
# find odd or even 

def findOddOrEven(params):
    str=""
    if params % 2 == 0:
        str=f"The number {params} is even"
    else:
        str=f"The number {params} not is even"

    return str
```

```python
result=findOddOrEven(7)
print(result)
```

```python
def allNames(*args):
   for i in args:
       print( f"Name: {i}")
   
allNames("shuvadeep", "rita", "debjyoti", "pulak")
 
```

```python
# keyword arguments:
def printDetails(**kwargs):
    for key, val in kwargs.items():
        print(f"{key}: {val}")

printDetails(name="shuvadeep", ph="7008485768", addrss="bbsr")
```

```python
def anyArgument(*args, **kwargs):
    for val in args:
        print(f"Position argument: {val}")

    for key, val1 in kwargs.items():
        print(f"{key}: {val1}")

anyArgument(1,2,3,4,5,6, name="sshuva", ph="9999999999")
```
