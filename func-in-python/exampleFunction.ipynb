#### Example 1: Temperature Conversion

```python
def convert_temperature(temp, unit):
    """This function converts temperature between celsius and fahrenheit"""
    if unit == "C":
        return temp * 9/5 + 32  ## celsius to fahrenheit
    elif unit == "F":
        return (temp-32) * 5/9
    else:
        return None

print(convert_temperature(26, "C"))
print(convert_temperature(89, "F"))
    
```

#### Example 2: Password strength checker

```python
def is_password_strong(pwd):
 
    if len(pwd) <= 8: 
        return False
    if not any(not char.isalnum() and not char.isspace() for char in pwd):
        return False
    if not any(char.isdigit() for char in pwd):
        return False
    return True     
 
print(is_password_strong(input("enter your password")))
# print(is_password_strong("shuva@deep1"))
# print(is_password_strong("!@#$%^&*$%"))
# print(is_password_strong("shuvadeep12"))
```

#### Emample 3: Calculate the total cost of items in a shopping cart

```python
def calculate_total_cost(cart):
    # pass
    total_cost=0
    for itm in cart:
        total_cost += itm['price'] * itm['qtn']

    return total_cost

cart = [
    {
        "name": "apple",
        "qtn": 7,
        "price": 90000 #630,000
    },
    {
        "name": "micromax",
        "qtn": 25,
        "price": 10000 #250,000
    },
    {
        "name": "samsung",
        "qtn": 4,
        "price": 190000 # 760,000
    }
    # total : 
]
# print(type(cart))
res=calculate_total_cost(cart)
print(res)
```

#### Example 4: Check IF a String is palindrome

```python
def is_palindrome(param):
    
    inputData = len(param)
    rev=""

    for i in range(inputData-1, -1, -1):
        rev += param[i]

    output_str =""
    if rev==param:
        output_str  = rev + " is palindrome"
    else:
        output_str  = rev + " is not palindrome"

    return output_str 

result = is_palindrome("515")
print(result)
```

#### Example 5: Calculate the factorials of a number using recursion

```python
def fectorial(param):
    if param == 0:
        return 1
    else: 
        return param * fectorial(param -1)

print(fectorial(5))

    
```
