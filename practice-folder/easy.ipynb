```python
## Ek function likho jo ek list leke usme se sabse bada aur sabse chhota number return kare (bina max()/min() use kiye).
def get_min_and_max(param):

    largest_numb=param[0]
    lowest_numb=param[0]

    for num in param:
        if num > largest_numb:
            largest_numb=num

        if num < lowest_numb:
            lowest_numb=num

    return largest_numb, lowest_numb
    
numbers = [45, 12, 89, 3, 67, 21]
small, big = get_min_and_max(numbers)
print("small : ", small)
print("big : ", big)
```

```python
## Ek list of strings diya hai — ["apple", "Banana", "cherry", "Date"]. Lambda + filter use karke sirf wo strings nikalo jo capital letter se start hote hain.

words = ["apple", "Banana", "cherry", "Date"]

get_capital_letter = list(filter(lambda word: word[0].isupper(), words))
print(get_capital_letter)
```

```python
## Ek tuple (10, 20, 30, 40, 50) diya hai — iska sum aur average nikalo.

numbers = (10, 20, 30, 40, 50)
total_sum = sum(numbers)
average = total_sum / len(numbers)

print("sum : ", total_sum)
print("average : ", average)
```
