```python
## Ek BankAccount class banao jisme __balance private variable ho, aur deposit(), withdraw(), get_balance() methods ho (getter/setter concept use karke). Withdraw karte waqt insufficient balance check karo.

class BankAccount:
    def __init__(self, insufficient_balance=0):
        self.__balance=insufficient_balance  # private attribute

    # Getter method:
    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance+=amount
            print(f"Deposited: {amount}. New Balance: {self.__balance}")
        else:
            print("Deposit amount 0 se zyada hona chahiye.")

    # Setter method 2: 
    def withdraw(self, amount):
        if amount <= 0:
            print("Withdraw amount 0 se zyada hona chahiye.")
        elif amount > self.__balance:
            print("Error: Insufficient balance! Transaction failed.")
        else:
            self.__balance-=amount

## Create object:
account = BankAccount(1000)
# print(account.get_balance()) # 1000

# account.deposit(500)
account.withdraw(2500)
account.withdraw(700)

print("Final Balance:", account.get_balance())





```
